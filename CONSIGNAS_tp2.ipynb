```python

# todo: Resolución
#Cuenta la frecuencia con la que se repiten las palabre de la cadena de texto
def contarFrecuencia(cadena):  
    palabras = cadena.split()  
    
    frecuencia = {}  
    
    for palabra in palabras:  
        if palabra in frecuencia:  
            frecuencia[palabra] += 1  
        else:  
            frecuencia[palabra] = 1  
            
    return frecuencia  

#Encuentra la palabra más repetida
def palabraMasRepetida(frecuencia):  
    palabraMax = max(frecuencia, key=frecuencia.get)  
    return (palabraMax, frecuencia[palabraMax]) 

#Pide una cadena de texto
cadena = input("Ingresa una cadena de caracteres: ")  

#imprime las palabras, su frecuencia y la palabra más repetida
diccionario = contarFrecuencia(cadena)  
print("Frecuencia de palabras:", diccionario)  
palabraFrecuencia = palabraMasRepetida(diccionario)  
print("La palabra más repetida es:", palabraFrecuencia)  
# ! 3
```

- **Actividad 2**: Escribir una función que reciba un número entero positivo y devuelva su factorial.


```python

# todo: Resolución
#Calcula la factura mas el iva 
def calcularFactura(cantidad, porcenIva=21):    
    iva = cantidad * (porcenIva / 100)  
    total = cantidad + iva  
    return total  

cantidad = float(input("Ingrese la cantidad sin IVA: "))  
porcentajeIva = float(input("Ingrese el porcentaje de IVA (o presione Enter para usar 21%): ") or 21)  

#Imprime el total mas el iva dependiendo de si se uso el iva por defecto o uno personalizado
if(porcentajeIva == 21):
  ivaDefault = calcularFactura(cantidad)  
  print(f"Total de la factura con el IVA al 21%): {ivaDefault}")  
else:
  ivaPersonalizado = calcularFactura(cantidad, porcentajeIva)  
  print(f"Total de la factura (incluyendo IVA al {porcentajeIva}%): {ivaPersonalizado}")  
# ! 1
```

```python

# todo: Resolución
#Calcula el factorial del numero
def calcularFactorial(n):  
    if n < 0:  
      print("El número debe ser un entero positivo")
    else:
      resultado = 1  
      for i in range(2, n + 1):  
          resultado *= i  
      return resultado  

#Pide un numero y llama a la funcion para calcular el factorial
numero = int(input("Ingrese un número entero positivo: "))
factorial = calcularFactorial(numero) 
if(numero < 0 ):
  pass
else:    
  print(f"El factorial de {numero} es: {factorial}")  
# ! 2
```
